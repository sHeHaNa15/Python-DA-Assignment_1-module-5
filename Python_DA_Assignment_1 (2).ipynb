# Part 1: NumPy Array Operations

## 1. Create a 1D NumPy Array

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

## 2. Inspection and Properties

```python
print("Shape             :", temperatures_w1.shape)
print("Data type         :", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

## 3. Array Operations

**Celsius to Fahrenheit:** `F = (C * 9/5) + 32`

```python
temperatures_w1_f = (temperatures_w1 * 9 / 5) + 32
print("Fahrenheit:", temperatures_w1_f)
```

```python
print("Maximum:", temperatures_w1.max())
print("Minimum:", temperatures_w1.min())
print("Mean   :", temperatures_w1.mean())
```

## 4. Array Slicing and Indexing

```python
print("First three days :", temperatures_w1[:3])
print("Weekend (last 2) :", temperatures_w1[-2:])
print("Middle three days:", temperatures_w1[2:5])
```

## 5. Create a 2D Array

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1],  # Week 2
])
print(temperatures)
```

## 6. Inspect and Slice the 2D Array

```python
print("Shape         :", temperatures.shape)
print("Data type     :", temperatures.dtype)
print("Total elements:", temperatures.size)
```

```python
print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])
print("\nWeekend (last two days) for both weeks:")
print(temperatures[:, -2:])
print("\nWeekend Week 1:", temperatures[0, -2:])
print("Weekend Week 2:", temperatures[1, -2:])
```

# Part 2: Pandas Series

## 1. Creating a Pandas Series

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"],
)
print(marks)
```

## 2. Indexing and Slicing

```python
#Access 1st rank student by integer position
print("1st rank mark:", marks.iloc[0])
```

```python
# loc accessor: top 3 ranks by label
print(marks.loc["Rank1":"Rank3"])
```

```python
#iloc accessor: 3rd rank student
print("3rd rank mark:", marks.iloc[2])
```

```python
#Boolean mask: marks greater than 90
above_90 = marks[marks > 90]
print(above_90)
print("Ranks:", list(above_90.index))
```

## 3. Manipulating Series

```python
#Modify the 1st rank mark to 100
marks["Rank1"] = 100
print(marks)
```

```python
#Remove the last rank entry
marks = marks.drop("Rank5")
print(marks)
```

```python
#Compute CGPA (marks / 10)
cgpa = marks / 10
print(cgpa)
```

# Part 3: Pandas DataFrame

## 1. Creating a Pandas DataFrame

```python
transactions = pd.DataFrame({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": ["Electronics", "Clothing", "Electronics", "Furniture",
                        "Clothing", "Electronics", "Furniture", "Clothing",
                        "Furniture", "Electronics"],
    "Region": ["North", "South", "North", "East", "West",
               "North", "East", "West", "South", "North"],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400],
})
transactions
```

## 2. Data Exploration

```python
transactions.head()
```

```python
transactions.tail()
```

```python
print("Shape:", transactions.shape)
print("Column names:", list(transactions.columns))
print("\nData types:")
print(transactions.dtypes)
```

```python
transactions.info()
```

```python
#Only ProductCategory and Amount
transactions[["ProductCategory", "Amount"]]
```

```python
#Last 3 columns using iloc
transactions.iloc[:, -3:]
```

```python
#Last 3 columns using loc
transactions.loc[:, ["ProductCategory", "Region", "Amount"]]
```

```python
#Region is North and Amount > 200
transactions[(transactions["Region"] == "North") & (transactions["Amount"] > 200)]
```

```python
#Value counts for ProductCategory
transactions["ProductCategory"].value_counts()
```

```python
#Unique values in Region
print(transactions["Region"].unique())
```

```python
#Mean amount per region
transactions.groupby("Region")["Amount"].mean()
```

## 3. Manipulating the DataFrame

```python
#Set Amount to 165 for TransactionID 102
transactions.loc[transactions["TransactionID"] == 102, "Amount"] = 165
transactions
```

```python
# Add Discount column 10%
transactions["Discount"] = transactions["Amount"] * 0.10
transactions
```

```python
#Remove the row with TransactionID 109
transactions = transactions[transactions["TransactionID"] != 109]
transactions
```

```python
#Delete the Discount column
transactions = transactions.drop(columns="Discount")
transactions
```
